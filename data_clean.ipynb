## 4. Data Cleaning

##### 4.1 Used name cleaning

```python
bios['Used name_clean'] = (
        bios['Used name']
        .str.replace('•', ' ')
        .str.replace(r'\s+', ' ')
        .str.strip()
)
```

##### 4.2 Full name cleaning

```python
bios['Full name_clean'] = (
    bios['Full name']
    .str.replace('•', ' ')
    .str.replace(r'\s+', ' ')
    .str.strip()
)
```

##### 4.3 Measurements Cleaning

```python
bios['Height_cm'] = np.nan
bios['Weight_min_kg'] = np.nan
bios['Weight_max_kg'] = np.nan
```

```python
height = bios['Measurements'].str.extract(
    r'(\d+(?:[.,]\d+)?)\s*cm',
    expand=False
)

bios['Height_cm'] = height.str.replace(',', '.', regex=False).astype(float)
```

```python
weight_text = bios['Measurements'].str.extract(
    r'(?:/\s*)?([\d\s,–-]+)\s*kg',
    expand=False
)
```

```python
def parse_weight(value):
    if pd.isna(value):
        return np.nan, np.nan

    value = value.strip()

    # Ondalıklı kilo: 77,5
    if re.fullmatch(r'\d+,\d+', value):
        weight = float(value.replace(',', '.'))
        return weight, weight

    # Aralık: 62–65 veya 62-65
    if re.fullmatch(r'\d+\s*[–-]\s*\d+', value):
        numbers = [float(x) for x in re.findall(r'\d+', value)]
        return min(numbers), max(numbers)

    # Çoklu kilo: 74, 82 veya 76, 77, 79
    numbers = re.findall(r'\d+', value)

    if numbers:
        numbers = [float(x) for x in numbers]
        return min(numbers), max(numbers)

    return np.nan, np.nan
```

```python
weights = weight_text.apply(parse_weight)

bios['Weight_min_kg'] = weights.str[0]
bios['Weight_max_kg'] = weights.str[1]
```
